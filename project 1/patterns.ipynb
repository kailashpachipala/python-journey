```python
for i in range (5):
    for j in range(6):
        print("*", end=" ")
    print()
```

```python
for i in range (5):
  
    print((i+1) * " *")   # time complexity is 0(n)
```

```python
for i in range (5):
    for j in range(i + 1):
        print("*", end=" ")      # time complexity is 0(n^2)
    print()
```

```python
for i in range (1, 6):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
for i in range (1, 6):
    for j in range(i):
        print(i, end=" ")
    print()
```

```python
for i in range (1, 6):
    for j in range(i):
        print(j, end=" ")
    print()
```

```python
for i in range (5, 0, -1):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
for i in range (5, 0, -1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
for i in range (5, 0, -1):
    for j in range(5, 5-i,-1):
        print(j, end=" ")
    print()
```

```python
for i in range (5 , 0, -1):
    for j in range(i):
        print(i, end=" ")
    print()
```
